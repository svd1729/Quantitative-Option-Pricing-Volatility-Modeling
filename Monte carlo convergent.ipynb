```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

# Import your existing classes
from black_scholes import BlackScholesPricer
from Monte_Carlo_Simulation_Engine import MonteCarloPricer
from Options_architecture import EuropeanOption, OptionType
```

```python
# ============================================================
# MARKET PARAMETERS
# ============================================================

S0 = 100.0
K = 105.0
T = 1.0
r = 0.05
sigma = 0.20

# European Call
option = EuropeanOption(
    strike=K,
    maturity=T,
    option_type=OptionType.CALL
)

# ============================================================
# BLACK-SCHOLES BENCHMARK
# ============================================================

bs_pricer = BlackScholesPricer(
    spot=S0,
    rate=r,
    volatility=sigma
)

bs_price = bs_pricer.price(option)

print(f"Black-Scholes Price: ${bs_price:.6f}")


# ============================================================
# MONTE CARLO CONVERGENCE
# ============================================================

mc_pricer = MonteCarloPricer(
    spot=S0,
    rate=r,
    volatility=sigma
)

# Number of simulation paths
path_counts = [
    1_000,
    2_500,
    5_000,
    10_000,
    25_000,
    50_000,
    100_000,
    250_000
]

results = []

for n_paths in path_counts:

    start_time = time.time()

    mc_price = mc_pricer.price(
        option=option,
        num_paths=n_paths,
        num_steps=252,
        antithetic=False,
        seed=42
    )

    runtime = time.time() - start_time

    absolute_error = abs(mc_price - bs_price)

    relative_error = (
        absolute_error / bs_price
    ) * 100

    results.append({
        "Paths": n_paths,
        "MC Price": mc_price,
        "BS Price": bs_price,
        "Absolute Error": absolute_error,
        "Relative Error (%)": relative_error,
        "Runtime (sec)": runtime
    })


results_df = pd.DataFrame(results)

display(results_df)
```
